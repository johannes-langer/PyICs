```python
%%writefile solutions/solution.py
import operator


def neighbors(graph: list[tuple[str, str, int]], node: str) -> list[str]:
    """Return the successors of a node."""
    return sorted(map(operator.itemgetter(1), filter(lambda x: x[0] == node, graph)))


def depth_first_search(
    graph: list[tuple[str, str, int]], start: str, goal: str
) -> list[str] | None:
    """Return a path from start to goal using depth-first search."""
    stack = []
    stack.append((start,))

    while stack:
        path = stack.pop()
        node = path[-1]

        if node == goal:
            return path

        for neighbor in reversed(neighbors(graph, node)):
            if neighbor in path:
                continue
            stack.append(path + (neighbor,))

    return None


if __name__ == "__main__":
    start = "A"
    goal = "G"
    undirected = True

    nodes = ["A", "B", "C", "D", "E", "F", "G"]
    edges = [
        ("A", "B", 4),
        ("A", "E", 10),
        ("B", "C", 5),
        ("B", "D", 2),
        ("B", "E", 6),
        ("D", "E", 1),
        ("D", "F", 3),
        ("E", "F", 6),
        ("E", "G", 1),
    ]
    all_edges = [(a, b, c) for a, b, c in edges] + [
        (b, a, c) for a, b, c in edges if undirected
    ]

    print(depth_first_search(all_edges, start, goal))
```
